## Implement the predict method

### Subtask:
Add a `predict` method that takes new data as input and assigns each data point to the closest cluster.


**Reasoning**:
Implement the predict method to assign new data points to the closest cluster based on the fitted centroids.



```python
class KMeans:
    def __init__(self, n_clusters=3, max_iter=100):
        self.n_clusters = n_clusters
        self.max_iter = max_iter
        self.centroids = None

    def fit(self, X):
        # 1. Initialize centroids randomly from the input data X
        n_samples, n_features = X.shape
        random_indices = np.random.choice(n_samples, self.n_clusters, replace=False)
        self.centroids = X[random_indices]

        # 2. Implement a loop that iterates up to self.max_iter
        for _ in range(self.max_iter):
            # 3. Assign each data point to the closest centroid
            distances = np.sqrt(((X - self.centroids[:, np.newaxis])**2).sum(axis=2))
            cluster_assignments = np.argmin(distances, axis=0)

            # 4. Recalculate the centroids
            new_centroids = np.zeros_like(self.centroids)
            for i in range(self.n_clusters):
                points_in_cluster = X[cluster_assignments == i]
                if len(points_in_cluster) > 0:
                    new_centroids[i] = points_in_cluster.mean(axis=0)
                else:
                    # If a cluster is empty, keep the old centroid
                    new_centroids[i] = self.centroids[i]

            # 5. Check for convergence
            if np.array_equal(new_centroids, self.centroids):
                break

            # Update centroids
            self.centroids = new_centroids

    def predict(self, X):
        # 1. Calculate the Euclidean distance from each data point in X to each of the fitted centroids
        distances = np.sqrt(((X - self.centroids[:, np.newaxis])**2).sum(axis=2))
        # 2. For each data point, find the index of the centroid that is closest to it.
        cluster_assignments = np.argmin(distances, axis=0)
        # 3. Return an array containing the cluster assignments
        return cluster_assignments
```
